```python

import pandas as pd

data = {
    'Student': ['Arun', 'Priya', 'Kavin', 'Divya', 'Rahul'],
    'Maths': [85, 90, 65, 95, 70],
    'Python': [88, 92, 70, 98, 75],
    'Science': [80, 85, 60, 90, 72],
    'Attendance': [90, 95, 75, 98, 80],
    'Study_Hours': [3, 4, 2, 5, 2]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("First 5 rows:")
print(df.head())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns)
```

```python
df['Average'] = (df['Maths'] + df['Python'] + df['Science']) / 3

print(df[['Student', 'Average']])
```

```python
topper = df.loc[df['Average'].idxmax()]

print("Topper Details:")
print(topper)
```

```python
# Calculate the class average

class_average = df['Average'].mean()

print("Overall Class Average:", round(class_average, 2))
```

```python
def performance_category(avg):
    if avg >= 90:
        return "Excellent"
    elif avg >= 80:
        return "Very Good"
    elif avg >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df['Performance'] = df['Average'].apply(performance_category)

print(df[['Student', 'Average', 'Performance']])
```

```python

performance_count = df['Performance'].value_counts()

print("Student Performance Distribution:")
print(performance_count)
```

```python

import matplotlib.pyplot as plt

# Create a bar chart
performance_count.plot(kind='bar')

# Add chart title and labels
plt.title("Student Performance Distribution")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")

# Display the chart
plt.show()
```

```python
print(df[['Student', 'Maths', 'Python', 'Science']])
```

```python
subject_average = df[['Maths', 'Python', 'Science']].mean()

print("Subject-wise Average Marks:")
print(subject_average)
```

```python

import matplotlib.pyplot as plt

subject_average.plot(kind='bar')

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python
print("Student Attendance Details:")
print(df[['Student', 'Attendance']])

print("\nAverage Attendance:")
print(df['Attendance'].mean())
```

```python

import matplotlib.pyplot as plt

subject_avg = df[['Maths', 'Python', 'Science']].mean()

plt.figure(figsize=(8,5))

plt.bar(subject_avg.index, subject_avg.values)

plt.title('Subject-wise Average Marks')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Study_Hours'], df['Average'])

plt.title('Study Hours vs Average Marks')
plt.xlabel('Study Hours')
plt.ylabel('Average Marks')

plt.show()
```

```python
correlation = df['Study_Hours'].corr(df['Average'])

print("Study Hours and Average Marks Correlation:")
print(correlation)
```

```python
highest = df['Average'].max()
lowest = df['Average'].min()

print("Highest Average Marks:", highest)
print("Lowest Average Marks:", lowest)
```

```python
df['Rank'] = df['Average'].rank(
    ascending=False,
    method='dense'
)

df = df.sort_values('Average', ascending=False)

print(df[['Student', 'Average', 'Rank']])
```

```python
df.to_csv('student_performance_analysis.csv', index=False)

print("Project data exported successfully!")
```
